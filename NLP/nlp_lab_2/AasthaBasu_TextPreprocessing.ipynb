# 1. Text Preprocessing with Pandas

```python
import pandas as pd

pd.set_option('display.max_colwidth', None)

df = pd.read_csv('childrens_books.csv')
df.head()
```

```python
df['cleaned_desc'] = df['Description'].str.lower()
df.head()
```

```python
df['cleaned_desc'] = df['Description'].str.replace(r'\[.*?\]', '', regex=True)
df.head()
```

```python
df['cleaned_desc'] = df['cleaned_desc'].str.replace(r'[^\w\s]', '', regex=True)
df.head()
```

## lower_replace()

```python
def lower_replace (series):
    output = series.str.lower()
    output = output.str.replace(r'\[.*?\]', '', regex=True)
    output = output.str.replace(r'[^\w\s]', '', regex=True)
    return output

lower_replace(df.Description)
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
phrase = df.cleaned_desc[0]
phrase
```

```python
doc = nlp(phrase)
doc
```

```python
norm = [token.lemma_ for token in doc if not token.is_stop]
norm
```

```python
list(nlp.Defaults.stop_words)[:10]
```

```python
' '.join(norm)
```

## token_lemma_nonstop()

```python
def token_lemma_nonstop(text):
    doc = nlp(text)
    output = [token.lemma_ for token in doc if not token.is_stop]
    output = ' '.join(output)
    return output

lower_replace(df.cleaned_desc).apply(token_lemma_nonstop)
```

# 2. POS Tagging

```python
phrase2 = lower_replace(df.cleaned_desc).apply(token_lemma_nonstop) [0]
phrase2
```

```python
doc2 = nlp(phrase2)
doc2
```

```python
pos = [(token.text, token.pos_) for token in doc2]
pos
```

```python
nouns = [(token.text) for token in doc2 if token.pos_ in ['NOUN', "PROPN"]]
nouns
```

```python
' '.join(nouns)
```

## filter_pos()

```python
def filter_pos(text, pos_list = ['NOUN', 'PROPN']):
    doc = nlp(text)
    output = [(token.text) for token in doc if token.pos_ in pos_list]
    output = ' '.join(output)
    return output

lower_replace(df.Description).apply(token_lemma_nonstop).apply(filter_pos)
```

# 3. Creating a Custom NLP Pipeline

```python
def nlp_pipeline(series):
    output = lower_replace(series)
    output = output.apply(token_lemma_nonstop)
    output = output.apply(filter_pos)
    return output

cleaned_text = nlp_pipeline(df.Description)
```

```python
pd.to_pickle(cleaned_text, 'cleaned_text.pkl')
```

# CountVectorizer

```python
series = pd.read_pickle('cleaned_text.pkl')
series
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
dtm = cv.fit_transform(series)
dtm_df = pd.DataFrame(dtm.toarray(), columns = cv.get_feature_names_out())
dtm_df
```

```python
cv2 = CountVectorizer(stop_words='english', ngram_range=(1, 2), min_df=2, max_df=.8)
cv2
```

```python
dtm2 = cv2.fit_transform(series)
dtm_df2 = pd.DataFrame(dtm2.toarray(), columns=cv2.get_feature_names_out())
dtm_df2
```

# 5. EDA with CountVectorizer

```python
term_freq = dtm_df2.sum()
term_freq
```

```python
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt

top_n = 30  # Show only the top 30 terms for clarity
top_terms = term_freq.sort_values(ascending=False).head(top_n)

fig = px.bar(x=top_terms.values, y=top_terms.index, labels={'x': 'Frequency', 'y': 'Term'}, title="Top 30 Term Frequencies")
fig.show()
```
