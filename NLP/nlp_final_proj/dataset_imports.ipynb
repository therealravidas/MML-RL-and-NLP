```python
from huggingface_hub import snapshot_download
import os

# Base directory to save your dataset
base_dir = "/home/BTECH_7TH_SEM/Desktop/peoples_speech_dirty_sa"

# Download only the 'dirty_sa' folder, keeping its subfolders (train/validation/test)
snapshot_download(
    repo_id="MLCommons/peoples_speech",
    repo_type="dataset",
    allow_patterns=[
        "dirty_sa/train/**",        # training data
        "dirty_sa/validation/**",   # validation data
        "dirty_sa/test/**",         # test data
        "dirty_sa/*",               # metadata files (like dataset_info.json)
    ],
    local_dir=base_dir
)

print(f"✅ Download complete. Files saved to:\n{os.path.abspath(base_dir)}")
```
