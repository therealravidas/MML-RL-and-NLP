```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, SimpleRNN, Dropout

# 1. Define Hyperparameters  
max_features = 20000  # Vocab size (number of most frequent words to keep)
maxlen = 100         # Max review length (truncate/pad reviews to this length)
batch_size = 64
embedding_dims = 50  # Dimension for the word embeddings
rnn_units = 32       # Number of units in the SimpleRNN layer

print("1. Loading data")

# 2. Load Data
# The data is conveniently pre-tokenized in Keras.
# Words are replaced by integers (their frequency rank).
# We limit the dataset to the top `max_features` most frequent words.
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

print(f"{len(x_train)} training sequences")
print(f"{len(x_test)} test sequences")
print("Data loaded.")

print("\n2. Preprocessing data (Padding sequences)")

# 3. Preprocess Data
# Reviews have different lengths. We must pad or truncate them to a fixed length (`maxlen`) for the RNN.

x_train = sequence.pad_sequences(x_train, maxlen=maxlen, padding='pre')
x_test = sequence.pad_sequences(x_test, maxlen=maxlen, padding='pre')

print(f"x_train shape: {x_train.shape}")
print(f"x_test shape: {x_test.shape}")
print("Data preprocessed.")

print("\n3. Building the RNN model")

# 4. Buildingh the Model
model = Sequential()

# 1. Embedding Layer:
# This layer turns the integer word indices into dense vectors of size 'embedding_dims'. It will be learned during training.
model.add(Embedding(input_dim=max_features,
                    output_dim=embedding_dims,
                    input_length=maxlen))
model.add(Dropout(0.25))

# 2. SimpleRNN Layer:
# The core recurrent layer that processes the sequence of word embeddings.
model.add(SimpleRNN(rnn_units))
model.add(Dropout(0.25))

# 3. Output Layer:
# A single neuron with a 'sigmoid' activation function.
# It outputs a probability (0 to 1) of the review being positive (class 1).
model.add(Dense(1, activation='sigmoid'))

# 5. Compile the Model
# We use 'binary_crossentropy' for binary (0/1) classification.
# 'adam' is a generally good optimizer.
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Print a summary of the model's architecture
model.summary()
print("Model built and compiled")

print("\n4. Training the model")

# 6. Train the Model
model.fit(x_train, y_train,
          epochs=50,
          batch_size=batch_size,
          validation_data=(x_test, y_test))

print("\n5. Evaluating the model")

# 7. Evaluate the Model
score, acc = model.evaluate(x_test, y_test, batch_size=batch_size)

print(f"\nTest score: {score:.4f}")
print(f"Test accuracy: {acc:.4f}")

print("\nModel training and evaluation complete")
```

# LSTM + RNN

```python
#!/usr/bin/env python3

"""
This script builds a Recurrent Neural Network (RNN) using an LSTM layer
to perform sentiment classification on the IMDB movie review dataset.
"""

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, LSTM, Dropout # Import LSTM

# --- 1. Define Hyperparameters ---
max_features = 20000  # Vocab size (number of most frequent words to keep)
maxlen = 100         # Max review length (truncate/pad reviews to this length)
batch_size = 64
embedding_dims = 50  # Dimension for the word embeddings
rnn_units = 32       # Number of units in the LSTM layer

print("1. Loading data")

# --- 2. Load Data ---
# The data is conveniently pre-tokenized in Keras.
# Words are replaced by integers (their frequency rank).
# We limit the dataset to the top `max_features` most frequent words.
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

print(f"{len(x_train)} training sequences")
print(f"{len(x_test)} test sequences")
print("Data loaded.")

print("\n2. Preprocessing data (Padding sequences)")

# --- 3. Preprocess Data ---
# Reviews have different lengths. We must pad or truncate them
# to a fixed length (`maxlen`) for the RNN.
# 'pre' padding adds 0s at the beginning of the sequence.
x_train = sequence.pad_sequences(x_train, maxlen=maxlen, padding='pre')
x_test = sequence.pad_sequences(x_test, maxlen=maxlen, padding='pre')

print(f"x_train shape: {x_train.shape}")
print(f"x_test shape: {x_test.shape}")
print("Data preprocessed.")

print("\n3. Building the LSTM model") # Updated title

# 4. Build the Model
model = Sequential()

# 1. Embedding Layer:
# This layer turns the integer word indices into dense vectors of
# size `embedding_dims`. It will be learned during training.
model.add(Embedding(input_dim=max_features,
                    output_dim=embedding_dims,
                    input_length=maxlen))
model.add(Dropout(0.25))

# 2. LSTM Layer:
# Using LSTM (Long Short-Term Memory) which is better at
# capturing long-range dependencies than SimpleRNN.
model.add(LSTM(rnn_units)) # Replaced SimpleRNN with LSTM
model.add(Dropout(0.25))

# 3. Output Layer:
# A single neuron with a 'sigmoid' activation function.
# It outputs a probability (0 to 1) of the review being positive (class 1).
model.add(Dense(1, activation='sigmoid'))

# 5. Compile the Model
# We use 'binary_crossentropy' for binary (0/1) classification.
# 'adam' is a generally good optimizer.
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Print a summary of the model's architecture
model.summary()
print("Model built and compiled.")

print("\n 4. Training the model")

# 6. Train the Model
model.fit(x_train, y_train, epochs=50, batch_size=batch_size, validation_data=(x_test, y_test))

print("\n5. Evaluating the model")

# 7. Evaluate the Model
# Check the model's performance on the unseen test data.
score, acc = model.evaluate(x_test, y_test, batch_size=batch_size)

print(f"\nTest score: {score:.4f}")
print(f"Test accuracy: {acc:.4f}")

print("\nModel training and evaluation complete")
```
