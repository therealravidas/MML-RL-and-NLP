```python
import random

class MENACE:
    def __init__(self): self.boxes = {}; self.history = []

    def move(self, b):
        s = "".join(b)
        moves = [i for i, v in enumerate(b) if v==" "]
        if s not in self.boxes: self.boxes[s] = {m:3 for m in moves}
        box = self.boxes[s]
        m = random.choices(list(box), weights=box.values())[0]
        self.history.append((s, m))
        return m

    def reward(self, result):
        for s, m in self.history:
            if result=="win":   self.boxes[s][m] += 3
            if result=="draw":  self.boxes[s][m] += 1
            if result=="loss":  self.boxes[s][m] = max(1, self.boxes[s][m] - 1)
        self.history=[]

def win(b):
    L=[(0,1,2), (3,4,5), (6,7,8), (0,3,6), (1,4,7), (2,5,8), (0,4,8), (2,4,6)]
    for i, j, k in L:
        if b[i] != " " and b[i]==b[j]==b[k]: return b[i]
    return "draw" if " " not in b else None

def train(n=3000):
    M=MENACE()
    for _ in range(n):
        b=[" "] * 9; p="X"
        while True:
            m = M.move(b) if p=="X" else random.choice([i for i, v in enumerate(b) if v == " "])
            b[m] = p
            r = win(b)
            if r:
                M.reward("win" if r == "X" else "loss" if r == "O" else "draw")
                break
            
            p = "O" if p == "X" else "X"
    return M

def play(M):
    b = [" "]*9; p = "X"
    pr = lambda:print(f"\n{b[0]}|{b[1]}|{b[2]}\n-----\n{b[3]}|{b[4]}|{b[5]}\n-----\n{b[6]}|{b[7]}|{b[8]}\n")
    while True:
        pr()
        m = M.move(b) if p == "X" else int(input("Your move 0-8: "))
        if p == "O" and b[m] != " ": print("Bad"); continue
        b[m] = p
        r = win(b)
        if r:
            pr()
            print("MENACE wins!" if r == "X" else "You win!" if r == "O" else "Draw")
            M.history = []
            return
        p = "O" if p == "X" else "X"

if __name__=="__main__":
    M = train(10000)
    play(M)
```
